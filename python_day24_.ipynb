## how to create our own exception

#### we can use class for creating own exception , and then we can give our own name in exception.and then we use exception class to inherit the behaviour from exception . after we create the own exception.we are must raise the exception .

```python
class loveerror(Exception):
    pass

try:
    age=int(input())

    if age < 19:
        raise loveerror("Your age should be greater then 19")

except loveerror as l:
    print(l)
    
```
