### Ternary Operater

#### a ternary operater having if and else in single line.if the given conditions is satisfy executed the left side ,if not satisfy execute the right side.

result1 if condition else result2

```python
age=10
print("You are eligible") if age>19 else print("Not eligible")
```

### define While Loop

#### a condition loop take python expression , if the given expression is generate true or true Value loop execute repeatly. once given expression generate false or false Value loop exit immediatly.

## while Loop syntax:

* 1.initialization
* 2.condition
* 3.increment

```python
n=int(input("Enter the Number:"))
i=1

while i<=n:
    if i%2==0:
        print("Even Number",i)
    i=i+1
    
    
    
```

### Multiplication Table example:

```python
n=int(input("Enter the Number:"))
i=1

while i<=10:
    print(n,"x",i,"=",n * i)
    i=i+1
```
