### 1. first we use  searching and replacing string methods

### 1.find() 

#### purpose: find() returns the position of the first occurrence of in a string. If the substring is not found, it returns -1.

```python
name="Neymar Jre"
```

```python
print(name.find("J"))
print(name.find("r"))
print(name.find("o"))
```

### rfind - rfind() returns the position of the last occurrence of  in a string. If the substring is not found, it returns -1..

```python
print(name.rfind("e"))
print(name.rfind("N"))
print(name.rfind("o"))
```

###  2.index - index() returns the position of the first occurrence of in a string. If the substring is not found, it raises a ValueError.
###  rindex() returns the position of the last occurrence of in a string. If the substring is not found, it raises a ValueError

```python
print(name.index("r"))
print(name.index("o"))
```

### 3. Count-Returns the number of occurrences  in a string.

```python
print(name.count("e"))
print(name.count("o"))
```

### 4.Startswith()-Checks whether the string starts with the given character  and returns True or False.

```python
print(name.startswith("Neymar"))
print(name.startswith("O"))
print(name.startswith("N"))
```

### 5. Endwith() - Checks whether the string ends with the given character or text and returns True or False.

```python
print(name.endswith("e"))
print(name.endswith("O"))
```

### 6.replace()- replace a given charcter with another charcter and return the new string

```python
name="Neymar Jrs"
new_name=(name.replace("Jrs","Jr"))
new_name1=(name.replace(" ","-"))
print(new_name)
print(new_name1)

```

```python
name="Neymar Jr"
new_name2=(name.replace("o","-"))
print(new_name2)
```
