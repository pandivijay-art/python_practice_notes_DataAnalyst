### define LEGB concept (how python Workes)

##### Local - in the current indentation Block ,first python see the current indentation Block.

#### Enclosed- parent indentation of current Block

##### Global -code we write in Margin ,can be accessed everywhere.

#### Build-in - the Value that is predefined in function .

#### WHen we call tha any Variable ,python first will searck object first in Local indentation , if not fount ,then next will check enclosed indentation , then will check global indentaion ,and lastly search build in 

```python
x = "Global"

def outer():
    x = "Outer"

    def middle():
        x = "Middle"

        def inner():
            print(x)
            print(len("Python"))

        inner()

    middle()

x = "Updated Global"

outer()
print(x)
```

#### what is The purpose of Global Keyword

#### Convert a Local Variable into Global Variable in simple termes or we make the local Variable accessplity to Global. we use global keyword , first we convert the local to Global, and then assign Values.

```python
def cricket():
    global player
    player="VIrst"   
    print("----------")
cricket()
print(player)    
```

#### define decoraters

#### a decoraters is a Python function it will Take function as argument by doing that we Can modify or extract some code around the receiving argument and it will return the new functions. we use @ symbol for decoraters. it is a shortcut sysntax for sent the parameter to decorater.

```python
def modi(a):
    print("Pak is the best team in pakistan")
    return modi
    
@modi
def cricket():
    print("Pak is the Best team")

cricket()
```
