### Abstruction

#### That hide the implementation Details and only show thw essential detail Feautures to user.(simple terms hide the unneccessary Detail and only show the feaures)
* #### cann't be access directly
* #### child class - must be implement the abstuct class method.

```python
from abc import ABC,abstractmethod

class Bank(ABC):
    @abstractmethod
    def interest_rate(self):
        pass

class sbi(Bank):
    def interest_rate(self):
        print("SBI interest Rate is 7%")

class pri():
    def call(self):
        print("satheesdh")

obj=sbi()
obj.interest_rate()

p=pri(Bank)
```

```python
from abc import ABC,abstractmethod

class payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass

class creditcart(payment):
    def pay(self,amount):
        print("Credit Cart payment:",amount)
        

class upi(payment):
    def pay(self,amount):
        print("UPI payment:",amount)


c=creditcart()
c.pay(500)

u=upi()
u.pay(500)
```

```python
from multipledispatch import dispatch

class calculater():
    @dispatch(int,int)
    def add(self,a,b):
        return a +b
        
    @dispatch(int,int,int)
    def add(self,a,b,c):
        return a+b+c

    @dispatch(str,str)
    def add(self,a,b):
        return a+b

c=calculater()
c.add(10,20,30)
c.add(10,20)    
c.add("satheesh","pandi")
        
```

#### OPERATOR overriding

### changing the nature of operator, inbetween two objects of classes, we cant use opertor directly , but we use opertor overriding we can archive that. alwayes (execute the left side object)
### we are change the behaviour with in class and object (not full python)
(__add__,etc... that has already predefine special magical methodes )

```python
class student:
    def __init__(self,mark):
        self.mark=mark

    def __add__(self,number):
        return self.mark - number.mark

    def __sub__(self,second):
        return self.mark + second.mark

s1=student(90)
s2=student(100)

print(s1+s2)
print(s1-s2)
        
```
