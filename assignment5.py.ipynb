### python does not have Tuble Comprehastion , but we convert generted expression to tuple using tuple().

### example:
#### only print the even numbers in tuple comprehastion.

```python
result=tuple(i for i in range(10) if i%2 == 0)
print(result)
```

## nested for loop in tuple comprehastion

### example:

```python
numbers = ((1, 2), (3, 4), (5, 6),(10,2323))
```

```python
result=tuple(j for i in numbers for j in i )
print(result)
```

### Only print even numbers

```python
numbers = ((1, 2), (3, 4), (5, 6),(10,2323))
result=tuple(j for i in numbers for j in i if j%2==0 )
print(result)
```

## set Comprehation

### it is a short way to generate the set useing for loop and optional conditions>

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 6]
```

```python
answer={i*i for i in numbers }
print(answer)
```

### example for nested for loop in set compreghastion


```python
sets = [
    {2, 4, 6},
    {1, 3, 5},
    {4, 5, 7}
]
finaly={j for i in sets for j in i }
print(finaly)
```

#### set - only contains the hasable elements.(hasable means immutable elements) but set itself mutable 

## only print the odd number useing nested for loop in (set comprehastion)

```python
sets = [
    {2, 4, 6},
    {1, 3, 5},
    {4, 5, 7}
]
finaly={j for i in sets for j in i if j%2 != 0}
print(finaly)
```
