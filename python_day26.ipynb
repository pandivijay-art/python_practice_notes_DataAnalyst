#### What is inheritence

#### inheritence is a process of which a subclass acquires the properties and methodes of superclass.
(subclass can use the properties and methodes of superclass is called inheritence)
#### superclass
* the class whose properties and methodes inherite by another class is called superclass.
* (superclass is a class that provides properties and methoids to another class through inheritence)
#### subclass
* that class inherit from another class is called subclass.

```python
class vehicle():
    def __init__(self,name):
        self.brand=name
    def start(self):
        print("vehicle is started")
        print("name:",self.brand)

class car(vehicle):
    def __init__(self,number,brand):
        self.model=number
        self.brand=brand
    def device(self):
        print("car is driving")
v=vehicle("toyota")
v.start()
c=car(2019,"toyota")
c.device()

        
        
```

#### super() 

#### super() is used inside a subclass to access the superclass properties and methodes 
* we can use superclassname to access the particular superclass properties.

```python
class person():
    def __init__(self,name):
        self.name=name

    def show_name(self):
        print("name :",self.name)
class student(person):
    def __init__(self,number,name):
        self.rollnu=number
        super().__init__(name)
    def show_student(self):
        print("name is:",self.name)
        print("roll is:",self.rollnu)

s1=student(101,"satheesh")
s1.show_student()
```

```python
class Person:
    name = "Satheesh"
    age = 19

class College(Person):
    college_name = "ABC College"
    location = "Madurai"
    age=90

class Student(College):
    course = "Computer Science"
    roll_no = 101

    def show_details(self):
        print(Person.name)
        print(super().age)

s=Student()
s.show_details()
```
