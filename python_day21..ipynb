#### filter() 



#### fiter is a build in fuction , it takes formula as a first argument and it takes sequence as a second argument ,apply the formula for the given sequences if return true it will be filtered , otherwise not filted.

```python
numbers = [12, 15, 18, 21, 24, 27, 30, 33, 36, 40]

dic=filter(lambda x:x%3== 0 and x%6!=0,numbers)
print(list(dic))
```

filter(function , iterable)

```python
words = [
    "python",
    "SQL",
    "data",
    "MachineLearning",
    "AI",
    "programming",
    "cat",
    "developer"
]

result=filter(lambda x:len(x)>5 and x[0] == x[0].lower(),words)
print(list(result))
```

### MaP()

### map() is a Build in function , it takes formula as argument and it takes sequence as a second Argument map(formula,sequence)  
* it will not filter ,it just apply the formula into the sequence and produce the new sequence.

```python
va=map(lambda x:x>3,range(10))
print(list(va))
```

```python
va=filter(lambda x:x>3,range(10))
print(list(va))
```

## Differnce Betweem map and filter

* filter it will remove the false Values and return only filter Values
* map it will not remove , just produce the new Values 

```python
def check(a,b):
    return a+b
result=map(check,range(10),range(10))
print(list(result))
```

```python
re=map(lambda x,y : x+y , range(10),range(10))
print(list(re))
```
