### generater()

#### Generater is a special type of function , that use the yield keyword instead of return , that is used to produce values one at time ,while remembering its previous state. and it wont execute whole function
* in every time calling , yield is not start with beginning ,because while calling its remembering the previous state.so execute the pause state.

### diff between return and yield

#### return keyword also used to return or produce a Value , but after return terminate or completly end the loop.only one return will execute, while in yield produce a Value and its remember the previous state.so big positive is , so,it can restart when it is stopped.yield not stop just pause.

```python
def fun():
    return 1
    return 2

a=fun()
```

```python
def fun():
    yield 1     # remember the previous state , so it is restart when its stopped.
    yield 2

a=fun()
print(next(a))
print(next(a))
```

## Example

```python
def count_number(n):
    while n>0:
        yield n
        n=n-1
    
obj=count_number(5)
print(next(obj))
print(next(obj))
print(next(obj))
print(next(obj))
print(next(obj))
```

```python
def count_number(n):
    while n>0:
        return  n
        n=n-1
    
obj=count_number(5)
print(obj)
print(obj)
```

## chainmap()

### Chain map is a class it takes n number of dictionary as a argument and combine all dictionary into single View.(not merge )

```python
from collections import ChainMap

d1={"name":"satheeh","age":19}
d2={"class":12,"couse":"DSA"}

result=ChainMap(d1,d2)   # not merge into new dictionary ,just give the result in single view
print(result["name"])
print(result["class"])
```

### differnce between merge and single View

### combine two or more dictiony into single dictioanry, not observe the old dictionary , and it will create the new dictionary.but,in single view observe the old dictionary

```python
d1={"name":"satheeh","age":19}
d2={"class":12,"couse":"DSA"}
d3=d1|d2
print(d3)
d1["age"]=90
print(d1)
```
