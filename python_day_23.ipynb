### How do handle Image in open() function

##### rb - to read the image and return the binary format of image.

```python
content=open("pic.jpg",'rb')
small=content.read()
print(small[:100])
```

```python
news=open("new_pic",'wb')
news.write(small)

```

```python
from PIL import Image
img=Image.open(news)
display(img)
```

### why we use csv Module

##### this is one of the build in moldule , it used to handle the csv files.

```python
import csv
```

##### open() - we can use open() to open the csv file , but this this function read the content only text format.it cant convert the dict or list. but we use csv build in module , this moldule it helpes us to convert the text to dict or list

```python
import csv
content=open("power.csv",'r')
listfile=csv.reader(content)
for i in listfile:
    print(i)
```

```python
import csv
content=open("power.csv",'r')
listfile=csv.DictReader(content)
for i in listfile:
    print(i)
```

#### reader() - it is convert the easch row as list format.

#### DIctreader() - it takes the first row as key , and remaing rows as Values , and it finally return the text to dictionary format

### context Manager in file Handling|

#### 'with' - this keyword is used to create the contact Manager. we use this keyword ,open the module in the specific Name spaces.context Manager that Manage the statup and close the file automatically. 

* with in namspace we only Handle the file ,once we get out the name sapce this file automatically closed.

```python
with open("power.csv",'r') as file:
    sa=file.read()
    print(sa)
```
