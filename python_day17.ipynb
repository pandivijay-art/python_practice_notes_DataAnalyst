```python
two_dime=[[1,2,3],
         [4,5,6],
         [7,8,]]
index=0
while index<len(two_dime):
    row=two_dime[index]
    print("Count row",row)

    inner_index=0
    while inner_index<len(row):
        print(row[inner_index])
        inner_index+=1

    index+=1
    print("================")
```

```python
numbers = [[10, 20],
           [30, 40, 50],
           [60, 70, 80, 90]]

index =0

while index<len(numbers):
    row=numbers[index]
    print(row)
    inner_index=0
    while inner_index<len(row):
        print(row[inner_index])
        inner_index+=1
    index+=1
    print("========")
```

```python
fruites=("apple","Mango","Graphes")
rate=(10,32,343,2323,3443534)
price=(122,45,678,32423)
list(zip(fruites,rate,price))
for i,j,k in zip(fruites,rate,price):
    print(f"My fruites Name is{i},my rate is {j},My price is {k}")
```

### List Comprehastion() - handling control flow and loop in list and generating,(its a short way to create new list using for and if condition )

```python
result=[x*x for x in range(10) if x%2==0 ] # if we use only if (we write the syntax - first for condition and  after write  if.  )
print(result)
```

```python
result=[x*x if x%2==0  else 0 for x in range(10)]  # if we use if and else (we write the syntax first if condition and else after for loop )
print(result)
```

### nested for loop with list comprehastion

```python
numbers = [[10, 20],
           [30, 40, 50],
           [60, 70, 80, 90]]
ans=[element for row in numbers for element in row]
print(ans)
```

```python
numbers = [[10, 15, 20],
           [25, 30, 35],
           [40, 45, 50]]

result=[j for i in numbers for j in i if j%5 != 0]
print(result)
```
