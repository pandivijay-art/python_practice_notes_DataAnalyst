### Multiple Inheritense

#### one subclass that inherit from more then one superclass is calles multiple inheritense.

#### super() only used in inside a class methodes

```python
class person:
    name="sathesh"
    age=19

    def show_person(self):
        print(self.name)
        print(self.age)
class student():
    course="B.sc"

    def show_student(self):
        print(self.course)

class college_student(person,student):

    def show_details():
        print(super().course)


c=college_student()
c.show_person()
c.show_student()
    
```

### hybrid inheritence

#### A combination of more then one inheritence together will Make hybrid Inheritence.

```python
class person():
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show_person(self):
        print(self.name)
        print(self.age)

class student(person):
    def __init__(self,name,age,course):
        person.__init__(self,name,age)
        self.course=course
    def show_student(self):
        print(self.name)
        print(self.age)
        print(self.course)

class employee(person):
    def __init__(self,course,salary):
        student.__init__(course)
        self.salary=salary
        
    def show_employee(self):
        print(self.course)
         

class collegemember(student,employee):
    pass

c=collegemember("sathesh",19,"b.sc")
c.show_person()
c.show_employee()  
c.show_student()

    
        
```
