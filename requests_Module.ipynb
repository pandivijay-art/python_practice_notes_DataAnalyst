## headers

#### headers means is the additional information we are sent along with https request.

for example:

* you are crete one bank account,like username=satheesh,password=1323
* after login, server will generate the authenticate token for every user.
* so,you want to check Bank Balance, so this is privacy matter. dont allow to every one access your Bank.
* so we have unique token,that token we are sent to the server using Headers.

```python
import requests
```

```python
url = "https://dummyjson.com/users/add"
data={
    "firstName": "satheesh",
    "lastName":1234,
    "age":20
}

new_user=requests.post(url,json=data)
print(new_user.status_code)
d=new_user.json()
print(d["username"])
print(d["password"])
```

```python
url2= "https://dummyjson.com/auth/login"
login_data = {
    "username": "emilys",
    "password": "emilyspass"
}
new_user=requests.post(url2,json=login_data)
print(new_user.status_code)
data2=new_user.json()
us=data2["username"]
token=data2["accessToken"]
print("my token number is:",bool(token))
print("my username is:",us)
```

### bearer is a authentication schema , I am presenting the token to proove my authentication Sucess.

```python
url="https://dummyjson.com/auth/me"
token=data2["accessToken"]
he={
    "Authorization":f"Bearer {token}"
}

user=requests.get(url,headers=he)
print(user.json())
```

# Handling error means handle the run time error 

#### raise_for_status() means check the status and raise exception ,if my status return like 400,401,it will raise exception

```python
try:
    response=requests.get("https://jsonplaceholder.typicode.com/posts/999")
    response.raise_for_status()
except requests.exceptions.RequestException as e:
    print("something wrong:",e)
```

* diff between status_code and raise_for_status : major differnce is status code only return  the status, raise_for_sattus check the status and if error it throe error.

```python
try:
    response=requests.get("https://jsonplaceholder.typicode.com/posts/999")
    print(response.status_code)
except requests.exceptions.RequestException as e:
    print("something wrong:",e)
```

* Requestexception means basically handlie the requested related errors,generally it almost handle the all related request library exception.

#### inside request library we have exception module , that exception have diifernd errors handling methodes.requests.exceptions.RequestsDependencyWarning

### session()

* cookies means small information storing in our server.

* session is a way to maintain information about user across multiple requests.

* cookies.get_dict() - means get all cookies and return them as python dictionary.

## with session

```python
import requests
```

```python
session=requests.session()
url = "https://httpbin.org/cookies/set?session_id=ABC123"
ser=session.get(url)
print(ser.status_code)
print(session.cookies.get_dict())

## profile
profile_url = "https://httpbin.org/cookies"
ser2=session.get(profile_url)
print(ser2.status_code)
print(session.cookies.get_dict())
```

# without session

#### without session means we are creating two separte requests,so second one will return empty.

```python

url = "https://httpbin.org/cookies/set?session_id=ABC123"
ser=requests.get(url)
print(ser.status_code)
print(ser.json())

## profile
profile_url = "https://httpbin.org/cookies"
ser2=requests.get(profile_url)
print(ser2.status_code)
print(ser2.json())
```
