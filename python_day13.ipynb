### How can we delete a dictioanry

##### 1.pop() - pop has nessary argument ,will delete the particular key.and return the deleting items.

```python
d1={"name":90,"age":90,"village":9000}
d2={"name":9000,"vil":89}
d1.pop("age")
print(d1)
```

#### 2.popitem() - no argument ,pop item will remove the last key value pair. and return the deleting items.

```python
d1={"name":90,"age":90,"village":9000}
d2={"name":9000,"vil":89}
d2.popitem()
```

#### 3.del .dictobj["key"] - delete the entire dictioanry and dictobject.and also delete the specific key value.

```python
d1={"name":90,"age":90,"village":9000}
d2={"name":9000,"vil":89}
del d1
```

#### 4.Clear() - wipe out all the element and return empty

```python
d1={"name":90,"age":90,"village":9000}
d2={"name":9000,"vil":89}
d1.clear()
print(d1)
```
