## Userdict

#### userdict is a class provided by collections module, it allow us to customized userdict it means which methoid should execute and which method should not execute .
* also we can introduce our own userdict with own Behaviouir

#### in dictionary __setitem__ is a special method , it is used to adding and inserting Values into dictionary using key and Values.

```python
from collections import UserDict

class smartdict(UserDict):
    def __setitem__ (self,key,value):
        if key in self.data:
             print(f"Duplicate keys not allowed",key)
        else:
            self.data[key]=value

    def delete(self,key):
        if key not in self.data:
            print(f"{key} not found")
        else:
            del self.data[key]

    def show_keys(self):
        print(self.data.keys())
    def show_values(self):
        print(self.data.values())
dic=smartdict({"name":"satheesh","age":19})

```

```python
dic["name"]="paanda"
```

```python
dic.delete("age1")
dic
```

### checkbox()

#### that allow user to select or seselect on option , we are select one or more options also

```python
import tkinter as ter
from tkinter import ttk

# create window
create=ter.Tk()
create.title("Check botton practice")
create.geometry("300x300")

# 2.lable creation
lab=ter.Label(create,text="python GUI",font=("Arial",20))
lab.pack(pady=20)

#3.booloean creation

var=ter.BooleanVar()

# check status

def check_status():
    if var.get():
        lab.config(text="you selection is better")
    else:
        lab.config(text="first you select")
button=ter.Checkbutton(create,text="I agree",command=check_status,variable=var)
button.pack(pady=20)
   
create.mainloop()
```

#### stringvar() - stored the text Value and connect to the grapical  user interface

```python
import tkinter as mod
from tkinter import ttk

# first i will Create window
window=mod.Tk()
window.title("Radio button example")
window.geometry("400x300")

label1=mod.Label(window,text="select only one option",font=("Times New Roman",10),bg="Red")
label1.pack(pady=10)

# next , we will create string Var
var=mod.StringVar(value="option is not selected")

# then we are create label
label=mod.Label(window,text="select only one option",font=("Times New Roman",20),bg="yellow")
label.pack(pady=10)

# now, we are create function
def radio_change():
    label.config(text=f"you selected course is {var.get()}")
    
# then , we are create radiobutton widget
radio1=mod.Radiobutton(window,text="B.sc computer science",command=radio_change,variable=var,value="b.sc selected")
radio2=mod.Radiobutton(window,text="B.com",command=radio_change,variable=var,value="b.com")
radio3=mod.Radiobutton(window,text="B.tech",command=radio_change,variable=var,value="b.tech")
radio1.pack(pady=20)
radio2.pack(pady=20)
radio3.pack(pady=20)
window.mainloop()
```

#### grid and frame()

```python
import tkinter as kind
window=kind.Tk()
window.title("the frame example")
window.geometry("500x400")
frame=kind.Frame()
frame.pack()

lable1=kind.Label(frame,text="first name",font=("Arial",10))
lable1.grid(row=0,column=0)

userinput1=kind.Entry(frame)
userinput1.grid(row=0,column=1)

lable2=kind.Label(frame,text="Age",font=("Arial",10))
lable2.grid(row=1,column=0)

userinput2=kind.Entry(frame)
userinput2.grid(row=1,column=1)

lable3=kind.Label(frame,text="Village",font=("Arial",10))
lable3.grid(row=2,column=0)

userinput3=kind.Entry(frame)
userinput3.grid(row=2,column=1)

window.mainloop()
```
