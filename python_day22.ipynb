#### 1.'r' mode

### r mode is a immutable mode , which allow us to see the content only , we cannot modify or change the content.

```python
files=open("sathe.txt",'r')
print(files.read())    # read() -return the all contentes
```

```python
files=open("sathe.txt",'r')
print(files.readline())    #readline() - read the line by line 
  
```

```python
files=open("sathe.txt",'r')
print(files.readlines())   #read the all lines a file and return the list format.stored in the list format.
```

### 'w' mode

#### a file can be opened in the write mode using mode a 'w' , if the file have alredy existing it remove or overwrite the all content , if file haev no content , it create a new content and add the contentes . if  you use w mode , every time we should close the file 

```python
file=open("sathe.txt",'w')
file.write("The greatest Batsman is Me")
file.close()
```

```python
file=open("sathe.txt",'w')
file.writelines(["The greatest Batsman is Me\n","the best bowler ihn the world\n"])
file.close()
```

```python
file=open("sathe.txt",'r')
print(file.read())
file.close()
```

### 'a' mode

#### add the content at the end of the file, if file already contain content , it wont remove the content instaed of add the content in the end of the file. if file doesnot exist it create a new file and add the content .

```python
vapa=open("sathe.py",'a')
vapa.writelines(["my name is satheesh\n","iam from beautifull tamilnadu\n"])
vapa.close()
```

```python
file=open("sathe.py",'r')
print(file.read())
file.close()
```
