## 1.Shallow copy()

### 2.Deepcopy
* deepcopy completely secregate the one object to another object, point the diffend internla refernces ,that cannot affect the original object.

```python
 from copy import deepcopy
```

```python
a=[[90,433,4546],[2302,2332,443]]
b=deepcopy(a)
b[0][0]=999
print(b)
print(id(b))
print(a)
print(id(a))
```
