### How Keyword Argument Function Works?

#### a function , the directly takes Parameters while calling,and assign Values for those Parameters we called Keyword Argument Fuction .

```python
def add(a,b):
    print("Value is:",a)
    print("Value is:",b)
    return a+b,a-b,a*b,a%b
     
ans=add(a=10,b=20)    # we assign Values to  directly Parametrs.
print(ans)
```
