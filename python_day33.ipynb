# Multi threading

* #### Multi threading that allow us to create our personal threads and execute multiple task concurrently.
* what is Global interpreter Lock:
* #### that allow us to execute only one thread a time to execute python Bytecode
* #### there is no relationship with global variable

```python
import threading
import time

start=time.time()
def first():
    for i in range(5):
        print("satheesh")
        time.sleep(2)
        print("pandi")

def second():
    for i in range(5):
        print("virat")
        print("kholi")

f1=threading.Thread(target=first)
f2=threading.Thread(target=second)
f1.start()
f2.start()

end=time.time()
print("consumed time is:",end-start)
            
```

#### Thread.Start - start the thread execution

#### join() - makes one thread wait untill another thread finishes its work.

```python
import threading
import time

start=time.time()
def first():
    for i in range(5):
        print("satheesh")
        time.sleep(2)
        print("pandi")

def second():
    for i in range(5):
        print("virat")
        print("kholi")

f1=threading.Thread(target=first)
f2=threading.Thread(target=second)
f1.start()
f1.join()
f2.start()
f2.join()

end=time.time()
print("consumed time is:",end-start)
            
```

## Date Extracting

* #### we are extarct date from text usind multiple symbols

```python
text = """
Ravi joined the company on 15-06-2024.
His first project started on 2024/08/20.
He completed training on 25/12/2025.
The next meeting is on 2026-10-14.
His birthday is 10-05-2004.
"""
date=re.findall(r"\b(?:\d{2}[-/]\d{2}[-/]\d{4} | \d{4}[-/]\d{2}[-/]\d{2})\b",text)
print(date)
```
