### strftime() -  it is used to convert the time or Date object  into formatting string

* %a - return the day name with short form
* %B - return the month Name with full form
* %b - return the month name wiht short form
* %C - return the current Century
* %c -it represent the local time and Date

```python
to=date.today()
to
```

## timedelta() - it is used to represenmt the diffence Between two date and times ,and also we find next 5 days or before 5 days

```python
from datetime import timedelta,datetime
```

```python
today=datetime.now()
```

```python
days=timedelta(days=10)
today+days
```

```python
from datetime import date, timedelta

today = date(2026, 10, 12)

future_date = today +  timedelta(days=10)
print(future_date)
```

```python
from datetime import date, timedelta

today = date(2026, 10, 12)

future_date = today +  timedelta(30)
print(future_date)
```
